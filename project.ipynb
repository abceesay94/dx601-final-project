### Problem 8 (40 points)

Perform principal components analysis of the input columns.
Compute how much of the data variation is explained by the first half of the principal components.
Build a linear regression using coordinates computed from the first half of the principal components.
Compare the mean squared error of this model to the previous model.
Plot actual targets vs predictions again. 

This problem depends on material from week 13.

```python
# YOUR CODE HERE
X = iris[["sepal-length", "sepal-width", "petal-length", "petal-width"]]
y = iris["target"]

X_mean = X.mean()
X_std = X.std()
X_standardized = (X - X_mean) / X_std
X_for_pca = X_standardized

pca = PCA()
pca.fit(X_for_pca)
ex = pca.explained_variance_ratio_
print(ex)

n_features = X_for_pca.shape[1]
k = n_features // 2
variance_first_half = np.sum(ex[:k])
print(variance_first_half)

X_pca = pca.transform(X_for_pca)
X_pca_reduced = X_pca[:, :k]

model_pca = sklearn.linear_model.LinearRegression()
model_pca.fit(X_pca_reduced, y)

y_pred_pca = model_pca.predict(X_pca_reduced)
mean_squared_pca = np.mean((y - y_pred_pca) ** 2)
print(mean_squared_pca)

plt.figure()
plt.scatter(y, y_pred_pca)
plt.xlabel("Actual Target")
plt.ylabel("Predicted Target")
plt.show()

# After running PCA on the four input columns, I found that the first two components explained about the first half of the 
# overall variation. I used these two components to build a new linear regression model. The MSE for this PCA model 
# was 0.07, compared to 0.02 for the original model that used all four features. This comparison shows that 
# this model was inferior to the other.
```
